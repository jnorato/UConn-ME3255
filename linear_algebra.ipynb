## Matrix Operations

```julia
A = [1 2 3; 4 5 6; 7 8 9]; 
2.0 * A # Scalar multiplication
```

```julia
v = [1, 2, 3] # Vector
A * v # Matrix-vector multiplication
```

```julia
B = [1 0 -1; 2 3 4; 0.1 0.2 0.3];
C = A + B
```

```julia
M = A*B # Matrix-matrix multiplication
```

```julia
B*A # Matrix multiplication is not commutative
```

```julia
A' # Transpose of A
```

```julia
det(A) # Determinant of C
```

```julia
G = rand(3,3);
inv(G) # Inverse of G
```
